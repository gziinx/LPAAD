```python
def calcular_quartis(tempo):
    dados = sorted(tempo)
    n = len(dados)
    posicoes = [(n + 1) / 4, (n + 1) / 2, (3 * (n + 1)) / 4]

    quartis = []
    for posicao in posicoes:
        posicao_inteira = int(posicao)
        if posicao == posicao_inteira:
                valor_quartil = dados[posicao_inteira - 1]
        else:
            parte_decimal = posicao - posicao_inteira
            valor_inferior = dados[posicao_inteira - 1]
            valor_superior = dados[posicao_inteira]
            diferença = valor_superior - valor_inferior
            proporção = diferença * parte_decimal
            valor_quartil = valor_inferior + proporção
        quartis.append(valor_quartil)
    return quartis
```

```python
import statistics

tempo = [48,32,55,41,37,62,29,45,51,35,70]


tempo_medio = statistics.mean(tempo)
mediana = statistics.median(tempo)
quartil_1, quartil_2, quartil_3 = calcular_quartis(tempo)
amplitude = max(tempo) - min(tempo)
variancia = statistics.pvariance(tempo)
desvio_padrao = statistics.pstdev(tempo)

iqr = quartil_3 - quartil_1
print(f"TEMPO MÉDIO: {tempo_medio}")
print(f"MEDIANA: {mediana}")
print(f"Q1: {quartil_1}")
print(f"Q2: {quartil_2}")
print(f"Q3: {quartil_3}")
print(f"AMPLITUDE: {amplitude}")
print(f"VARIÂNCIA: {variancia}")
print(f"DESVIO PADRÃO: {desvio_padrao}")
print(f"VALORES: {tempo}")
print(f"IQR: {iqr}")
```
