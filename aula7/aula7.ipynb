```python
def registrar_ocorrencia (codigo, *sistemas, **informacoes):
    print(f"CODIGO: {codigo}\n")
    print(f"Quantidade de Sistemas afetados: {len(sistemas)}\n")
    print(f"SISTEMAS:")
    for indice, valor in enumerate(sistemas):
        print(f"{indice + 1}: {valor}")
    print(" ")    
    for indice, valor in informacoes.items():
        print(f"INFORMAÇÕES:\n{indice}: {valor}")

    
```

```python
registrar_ocorrencia("OSPAD", "PC 1", "PC 2", Windows= "Virus")
```

```python
def cadastrar_funcionario(nome, idade, salario, setor):
    setores = ["TI", "RH", "Financeiro", "Comercial"]
    if not nome:
        print("NOME É OBRIGATÓRIO")
        return
    if idade < 18 or idade > 71:
        print("Idade Inválida")
        return
    if salario < 0:
        print("Salário Inválido")
        return
    if setor not in setores:
        print("Setor Inválido")
        return
    print(f"NOME: {nome}\nIDADE: {idade}\nSALÁRIO: {salario}\nSETOR: {setor}")
```

```python
cadastrar_funcionario("Gustavo",18,1000,"TI")
print(" ")
cadastrar_funcionario("",18,1000,"TI")
print(" ")
cadastrar_funcionario("Gustavo",16,1000,"TI")
print(" ")
cadastrar_funcionario("Gustavo",18,-1,"TI")
print(" ")
cadastrar_funcionario("Gustavo",18,1000,"YOUTUBER")
```
